# Research Task

## Question 1: What is OCR (Optical Character Recognition)?

OCR (Optical Character Recognition) is a technology that enables computers to recognize text from images, scanned documents, or photographs and convert it into editable digital text. It helps automate data entry and reduces the need for manual typing. OCR is widely used in document digitization, banking, education, and business applications.

## Question 2: Why is Urdu OCR harder than English OCR?

Urdu OCR is more difficult than English OCR because Urdu is written from right to left and its letters change shape depending on their position in a word. Many Urdu characters look very similar, which makes it challenging for computers to recognize them correctly. Different fonts, handwriting styles, and connected letters also increase the complexity of Urdu OCR.

## Question 3: What are two real-world situations where Urdu OCR would be useful?

Urdu OCR can be used to digitize old Urdu books, newspapers, and historical documents so they can be searched and preserved digitally. It is also useful for government offices, schools, and libraries to convert printed records into digital databases. Another application is extracting text from Urdu signboards, forms, and invoices for automation and record keeping.
